```python
from sklearn.model_selection import train_test_split
import sentencepiece as spm
import pandas as pd
```

```python
ROOT_PATH = '/content/drive/MyDrive/Colab Notebooks'
data_frame = pd.read_csv(f'{ROOT_PATH}/split/std_dialect.csv', encoding='cp949')
```

```python
texts = [row["dialect"] for _, row in data_frame.iterrows()] + [row["standard"] for _, row in data_frame.iterrows()]

train_text, test_text = train_test_split(texts, test_size = 0.3, random_state = 999)
print(len(train_text), len(test_text))
```

```python
def write_text(texts, data_type = 'train'):
  dummy = ''
  for idx, text in enumerate(texts, 1):
    print(f'[{data_type}-{idx}] {text}')
    content = f'{text} \n\n\n\n' if data_type == 'train' else f'{text} \n'
    dummy += content
    
  dataset = open(f'{data_type}.txt', 'w')
  dataset.write(dummy)
```

```python
write_text(train_text)
write_text(test_text, data_type='test')
```

```python
corpus = f'/content/train.txt'
prefix = 'dialect'
vocab_size = 32000
```

```python
spm.SentencePieceTrainer.train(
    f'--input={corpus} --model_prefix={prefix} --vocab_size={vocab_size + 7}'+
    ' --model_type=bpe --max_sentence_length=999999 --character_coverage=1.0' +
    ' --pad_id=0 --pad_piece=[PAD] --unk_id=1 --unk_piece=[UNK]' +
    ' --bos_id=2 --bos_piece=[BOS] --eos_id=3 --eos_piece=[EOS]' +
    ' --user_defined_symbols=[SEP],[CLS],[MASK]'
)
```

```python
vocab_file = '/content/dialect.model'
vocab = spm.SentencePieceProcessor()
vocab.load(vocab_file)
```

```python
for idx, text in enumerate(test_text):
  pieces = vocab.encode_as_pieces(text)
  ids = vocab.encode_as_ids(text)

  print(f'[{idx}] \n원본 : {text} \n나눔 : {pieces} \n인덱스 : {ids} \n')
```
